```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
energy_ccsdt = -100.2061487621313
```

```python
df = pd.read_csv("hf_energy_vs_bond_dim.csv", index_col="i")
print(df.head())
```

```python
df["error"] = abs(df["energy"] - energy_ccsdt) / abs(energy_ccsdt)
print(df.head())
```

```python
fig, ax = plt.subplots()
ax.set_yscale("log")
sns.lineplot(ax=ax, data=df, x="chi", y="error", hue="d")
```
