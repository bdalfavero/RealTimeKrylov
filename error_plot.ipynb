```python
import h5py
import matplotlib.pyplot as plt
```

```python
f = h5py.File("data/subspace_errors.hdf5")
steps = f["steps"][:]
h_errs = f["h_errs"][:]
s_errs = f["s_errs"][:]
f.close()
```

```python
fig, ax = plt.subplots()
ax.set_yscale('log')
ax.set_xscale('log')
ax.plot(steps, h_errs, label = "H")
ax.plot(steps, s_errs, label = "S")
```
